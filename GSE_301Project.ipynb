```python
import os

# 1a & 1b: Reading and storing data
# 2a & 2b: Handling missing files and invalid values
numerical_data = []

try:
    if os.path.exists("marks.txt"):
        with open("marks.txt", "r") as file:
            for line in file:
                try:
                    # Convert to float and store
                    value = float(line.strip())
                    numerical_data.append(value)
                except ValueError:
                    print(f"Skipping invalid data: {line.strip()}")
    else:
        print("Error: File 'marks.txt' does not exist.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# Creating calculation functions
# Using loops and operators for manual calculation

def calculate_stats(dataset):
    total = 0
    count = 0
    minimum = dataset[0] if dataset else 0
    maximum = dataset[0] if dataset else 0

    for value in dataset:
        total += value        # Arithmetic Operator
        count += 1            # Counting for average
        if value < minimum:
            minimum = value   # Comparison logic
        if value > maximum:
            maximum = value

    average = total / count if count > 0 else 0
    return total, average, minimum, maximum
```

```python
# Task 3 & 4: Logic Functions using Loops and Arithmetic Operators
def calculate_metrics(data):
    if not data:
        return 0, 0, 0, 0

    total = 0
    count = 0
    minimum = data[0]
    maximum = data[0]

    # Task 4b: Loop to traverse dataset
    for value in data:
        total = total + value    # Task 4a: Arithmetic (+)
        count = count + 1        # Counting for average

        # Comparison logic for Min/Max
        if value < minimum:
            minimum = value
        if value > maximum:
            maximum = value

    average = total / count      # Task 4a: Arithmetic (/)
    return total, average, minimum, maximum
```
