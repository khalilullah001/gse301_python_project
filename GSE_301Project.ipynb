Function

```python
# Creating calculation functions
# Using loops and operators for manual calculation

def calculate_stats(dataset):
    total = 0
    count = 0
    minimum = dataset[0] if dataset else 0
    maximum = dataset[0] if dataset else 0

    for value in dataset:
        total += value        # Arithmetic Operator
        count += 1            # Counting for average
        if value < minimum:
            minimum = value   # Comparison logic
        if value > maximum:
            maximum = value

    average = total / count if count > 0 else 0
    return total, average, minimum, maximum
```

Object oriented programming

```python
import os

class DataSet:
    def __init__(self, numeric_file, category_file, threshold=50):
        self.numeric_file = numeric_file
        self.category_file = category_file
        self.threshold = threshold
        self.data = []
        self.categories = set()
        self.statistics = {}

    def load_data(self):
        # Load numeric data
        if os.path.exists(self.numeric_file):
            with open(self.numeric_file, "r") as file:
                for line in file:
                    line = line.strip()
                    if line.isnumeric():
                        self.data.append(float(line))
        else:
            print(f"Error: {self.numeric_file} not found.")

        # Load categorical data
        if os.path.exists(self.category_file):
            with open(self.category_file, "r") as file:
                for line in file:
                    category = line.strip()
                    if category:
                        self.categories.add(category)
        else:
            print(f"Error: {self.category_file} not found.")

    def calculate_statistics(self):
        if not self.data:
            print("No valid numeric data to calculate statistics.")
            return

        total = sum(self.data)
        average = total / len(self.data)
        minimum = min(self.data)
        maximum = max(self.data)
        performance = "High Performance" if average > self.threshold else "Needs Improvement"

        self.statistics = {
            "Total": total,
            "Average": average,
            "Minimum": minimum,
            "Maximum": maximum,
            "Performance": performance,
            "Unique Categories": len(self.categories)
        }

    def display_results(self):
        if not self.statistics:
            print("No statistics available.")
            return

        print("\n===== RESULTS =====")
        for key, value in self.statistics.items():
            print(f"{key}: {value}")
```
