```python
def check_eligibility(student_profiles, student_name):
    # Find the student profile
    student = next((s for s in student_profiles if s["name"] == student_name), None)

    if student is None:
        return False, f"Error: Student '{student_name}' not found."

    # Assume 0 outstanding courses means no outstanding courses
    # Add a key 'outstanding_courses' to the student dictionaries in Part 1.2
    cgpa = student.get("cgpa", 0.0) [cite: 63]
    outstanding = student.get("outstanding_courses", 0) # Placeholder: 0 means None [cite: 64]
    active = student.get("is_active", False) [cite: 65]

    # Eligibility check using logical operators (and) [cite: 61]
    is_eligible = (cgpa >= 2.5) and (outstanding == 0) and (active is True)

    message = f"Matric Number: {student['matric']}\n" + \
              f"CGPA: {cgpa}\n" + \
              f"Outstanding Courses: {outstanding}\n" + \
              f"Active Status: {active}\n" + \
              "Eligibility Result:"

    if is_eligible:
        message += f"\n{student_name} is eligible for graduation." [cite: 145]
    else:
        # Provide detailed reasons for ineligibility (can use 'or' logic here for messaging)
        reasons = []
        if cgpa < 2.5:
            reasons.append("CGPA is below the minimum 2.5 requirement.")
        if outstanding > 0:
            reasons.append(f"Outstanding courses count is {outstanding} (must be 0).")
        if not active:
            reasons.append("Student is not currently active.")

        message += f"\n{student_name} is NOT eligible for graduation. Reasons: {' and '.join(reasons)}"

    return is_eligible, message
print (check_eligibility)
```

```python
nested_scores = {
    "Yusuf Adeoye": {"Python": 85, "Statistics": 70, "Algorithms": 92},
    "Rasheed Fatia": {"Python": 95, "Statistics": 99, "Algorithms": 90},
    "Nimah Nina": {"Python": 60, "Statistics": 75, "Algorithms": 80}
}

def analyze_scores(scores_dict):
    student_averages = {}
    high_achievers = []

    for student, courses in scores_dict.items():
        # Calculate the average score for each student [cite: 70]
        total_score = sum(courses.values())
        num_courses = len(courses)
        average = total_score / num_courses
        student_averages[student] = round(average, 2)

        # Identify students who scored above 70 in all registered courses [cite: 71]
        all_above_70 = all(score > 70 for score in courses.values())
        if all_above_70:
            high_achievers.append(student)

    return student_averages, high_achievers
```
