```python
# =================================================================
# Colab Cell 4: Interactive Menu System and Execution (Part 4)
# =================================================================

# Task 4.2: Eligibility Checker (uses logical operators)
def check_eligibility(profiles, student_name):
    """Uses logical operators to determine graduation eligibility."""

    student = next((s for s in profiles if s["name"].lower() == student_name.lower()), None)

    if student is None:
        return False, f"Error: Student '{student_name}' not found."

    cgpa = student.get("cgpa", 0.0)
    outstanding_courses = student.get("outstanding_courses", -1)
    active_status = student.get("is_active", False)

    # Eligibility check using logical operators (`and`)
    is_eligible = (cgpa >= 2.5) and (outstanding_courses == 0) and (active_status is True)

    # Formatting output to match sample output
    message = f"Checking eligibility...\n" + \
              f"Matric Number: {student['matric']}\n" + \
              f"CGPA: {cgpa}\n" + \
              f"Outstanding Courses: {outstanding_courses}\n" + \
              f"Active Status: {active_status}\n" + \
              f"Eligibility Result:"

    if is_eligible:
        message += f"\n{student_name} is eligible for graduation."
    else:
        # Provide detailed reasons for ineligibility
        reasons = []
        if cgpa < 2.5:
            reasons.append("CGPA is below the minimum 2.5 requirement.")
        if outstanding_courses > 0:
            reasons.append(f"Outstanding courses count is {outstanding_courses} (must be 0).")
        if not active_status:
            reasons.append("Student is not currently active.")

        message += f"\n{student_name} is NOT eligible for graduation. Reasons: {' and '.join(reasons)}"

    return is_eligible, message

# Helper function for adding a student (uses Task 2.2 validation from Cell 2)
def add_new_student(profiles):
    """Prompts user for new student details and adds to the profile list."""
    print("\n--- Add New Student ---", flush=True)
    name = input("Enter name: ")
    matric = input("Enter matric number: ")

    age = get_validated_input("Enter age: ", int, 16, 40)
    cgpa = get_validated_input("Enter CGPA: ", float, 0.0, 5.0)

    active_str = input("Is the student active (yes/no): ").lower()
    active = True if active_str == 'yes' else False

    courses_str = input("Enter courses (comma separated): ")
    courses = [c.strip() for c in courses_str.split(',')]

    new_student = {
        "name": name,
        "matric": matric,
        "age": age,
        "cgpa": cgpa,
        "is_active": active,
        "courses": courses,
        "grades": {},
        "outstanding_courses": 0
    }

    profiles.append(new_student)
    print("Student record added successfully.", flush=True)
    print("--------------------------------------------", flush=True)

# Task 4.1: Build a Console Menu (WHILE loop and MATCH CASE)
def main_menu(profiles):
    """Creates a repeating console menu using a WHILE loop and MATCH CASE."""

    # Initial setup messages (Sample Output)
    print("============================================", flush=True)
    print("Student Academic Performance System", flush=True)
    print("============================================", flush=True)
    print("Loading student records...", flush=True)
    print(f"{len(profiles)} student profiles loaded successfully.", flush=True)
    print("--------------------------------------------", flush=True)

    while True: # Menu should repeat until the user selects Exit
        # Display Menu Options (Sample Output)
        print("\nMenu Options", flush=True)
        print("1. View all students", flush=True)
        print("2. Add new student", flush=True)
        print("3. Check eligibility for graduation", flush=True)
        print("4. Find top performer", flush=True)
        print("5. Exit", flush=True)
        print("--------------------------------------------", flush=True)

        choice = input("Enter your choice: ") # Execution pauses here

        # Use MATCH CASE for menu handling
        match choice:
            case '1':
                # View all students
                print("\nList of Students:", flush=True)
                for i, student in enumerate(profiles, 1):
                    print(f"{i}. {student['name']}", flush=True)
                print("--------------------------------------------", flush=True)

            case '2':
                # Add new student
                add_new_student(profiles)

            case '3':
                # Check eligibility for graduation
                student_name_check = input("Enter student name: ")
                is_eligible, message = check_eligibility(profiles, student_name_check)
                print(message, flush=True)
                print("--------------------------------------------", flush=True)

            case '4':
                # Find top performer
                top_student = max(profiles, key=lambda s: s['cgpa'])

                print("\nTop Performer:", flush=True)
                print(f"Name: {top_student['name']}", flush=True)
                print(f"Matric: {top_student['matric']}", flush=True)
                print(f"CGPA: {top_student['cgpa']}", flush=True)
                print(f"Courses: {top_student['courses']}", flush=True)
                print("--------------------------------------------", flush=True)

            case '5':
                # Exit
                print("Exiting the system...", flush=True)
                print("Goodbye!", flush=True)
                break # Exit the loop

            case _:
                print("Invalid choice. Please select a valid option (1-5).", flush=True)


# --- Main Execution Block ---
if __name__ == "__main__":
    # Start the interactive menu system
    main_menu(student_profiles)
```
